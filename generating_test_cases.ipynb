```python
import numpy as np
import pandas as pd
```

```python
np.random.seed(0)

random_points = np.random.rand(10, 3)
np.savetxt(f"point_sets/3D_uniform_{random_points.shape[0]}.csv", random_points, delimiter=",")
for _ in range(18):
    new_random_points = np.random.rand(5, 3)
    random_points = np.vstack([random_points, new_random_points])
    np.savetxt(f"point_sets/3D_uniform_{random_points.shape[0]}.csv", random_points, delimiter=",")
```
